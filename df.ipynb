Remover os processos duplicados, mantendo apenas os vencedores

```python
df_unicos = df_bruto.copy()
print('tamanho do dataset original:', df_unicos.shape)
df_unicos = df_unicos.drop_duplicates(subset=['Protocolo', 'Situação Proposta'], keep='first')
df_unicos = df_unicos[df_unicos['Situação Proposta'] == 'Vencedora']
print('tamanho após manter apenas os vencedores:', df_unicos.shape)
```
